```python
import re

def to_salce_lan(text, consonant="c"):
    """translate from Russian to salty language"""
    vowels = "аеёиоуыэюяАЕЁИОУЫЭЮЯ"
    result= ""

    for char in text:
        result+=char
        if char in vowels:
            result += consonant + char.lower()
    print(result)

#print(to_salce_lan("У попа была собака"))

def extract_chords(file_path): 
    """Extract all chords from songs file"""
    chords=set()
    pattern = r'\b[A-G][a-zbm#]*\b'

    with open(file_path, 'r') as file:
        for line in file:
            found=re.findall(pattern, line)
            for chord in found:
                chords.add(chord)
    return chords

#chords = extract_chords('data/song.txt')
#print(chords)


```
